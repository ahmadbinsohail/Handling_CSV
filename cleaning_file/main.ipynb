```python
import pandas as pd
import numpy as np
```

```python
from pathlib import Path as P
try :
    dir = P(__file__).resolve().parent
except NameError:
    dir = P.cwd()
```

```python
file_path = dir / "data" 
df = pd.read_csv(file_path /"scrapped_data.csv")
```

```python
try :
    df["Release Date"]=df["Release Date"].str.strip()   # .strip() for removel of hidden or extra spaces
    df["Release Date"]= pd.to_datetime(df["Release Date"],format= "mixed", errors="coerce")
except AttributeError :     # To control error on ipynb format if single cell run 
    df["Release Date"]= pd.to_datetime(df["Release Date"],format= "mixed", errors="coerce")
```

```python
# Deleting Fake dates
df["Release Date"]=pd.to_datetime(df["Release Date"], dayfirst=True)  #dayfirst=True format so easily read it
Fake_dates= df.loc[df["Release Date"].dt.year>2026,"Release Date"]
df.loc[df["Release Date"].dt.year>2026,"Release Date"] = pd.NaT
```

```python
# Checking how many units have 0.0 values (No sales)
shares_zero=(df["Units(m)"]==0).mean() * 100
print(f"{shares_zero:.1f} of games have 0.0 units")
```

```python
# Create Have_sales column for the True and false 
df["Have_sales"] = df["Units(m)"] > 0
# Cleaning the column Units(m) by making 0.0 units Nan values
df["Units(m)"]= df["Units(m)"].replace(0, np.nan)
```

```python
# Cleaning The Game Name Column
# 1st removing data contains testing
df = df[~df["Game Name"].str.contains("Testing", case=False)]
```

```python
# Deleting near duplicate rows
df = df.sort_values("Units(m)" ,ascending= False , na_position="last")
df = df.drop_duplicates(subset=["Game Name","System Full"], keep="first")
```
